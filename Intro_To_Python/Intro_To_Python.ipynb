## Variables & Syntax

```python
# We can assign variables with "="
x = 5
print(x)

# Variables are not constant: they
# can be overwritten or modified!
x = 6
print(x)

# here we assign the value True
# to a variable named condition.

# This essentially means we say
# condition is True
condition = True

if condition:
    print('This is indented')
```

### Python Variable Naming Convetions

**Hard rules:**
- Names can't start with a number (`2cool` is invalid, `cool2` is fine)
- No spaces or special characters like `-`, `!`, `@` (use `_` instead)
- Can't use reserved words like `if`, `for`, `return` as variable names

**Conventions:**
- Some languages use camelCase (`myVariableName`) — Python typically uses **snake_case** (`my_variable_name`)
- Constants (values that shouldn't change) are usually written in **ALL_CAPS**, like `MAX_SPEED = 120`

## Print and Input
`print()` displays something on the screen. `input()` pauses the program, asks the user to type something, and hands back whatever they typed.

```python
print("Hello!")
print("The answer is", 42)   # separate multiple things with commas

name = input("What's your name? ")
print(f"Nice to meet you, {name}!")
```

Note that `input()` always returns a string, even if the user types a number. That's a common source of bugs, and it leads into the next section.


## Type Conversion
Sometimes you need to turn a value from one type into another (like when you get an input, as a string, and want a number). The main converters are `int()`, `float()`, and `str()`.

```python
print(int('7'))      # 7
print(float('3.5'))  # 3.5
print(str(42))       # '42'
print(int(9.99))     # 9 (drops the decimal, doesn't round)
```

## Comparison Operators & Conditionals
`if` checks one condition. `elif` ("else if") checks another condition, but only if the first one was `False`. `else` runs if none of the conditions above it were `True`.

This also uses comparison operators: `>`, `<`, `==`, `!=`, `>=`, `<=`.

Notice how the assignment operator `=` is not the equality operator `==`.

```python
# assigning a variable
x = 5

# checking equality 
if x == 5:
    print('x is equal to 5')
```

```python
temperature = 75

if temperature > 90:
    print("It's hot")
elif temperature > 60:
    print("It's mild")
else:
    print("It's cold") # notice here that we use "" because we want to use ' inside the string
```

## For Loops
A `for` loop repeats a block of code once for each item in a collection. The indented lines run once per item.

```python
capitals = ['Tokyo', 'Kyiv', 'Cairo']

for capital in capitals:
    print(capital)
```

`range()` gives you a sequence of numbers to loop over. Like list indexing, it starts at 0.

```python
for i in range(5):
    print(i)
```

Python also has **while loops**, which keep repeating as long as a condition is `True`. They're useful when you don't know ahead of time how many times to repeat, but we won't need them today.

## Functions
A function is a reusable block of code that performs a specific task. Instead of rewriting the same lines every time you need to do something, you define it once and can call it as many times as you want, with different inputs.

```python
# functions

def greet(name):
    print("Hello,", name)

# enter your name
greet('Amy')
```

f-strings are formed like this: `f""`. Variables can be referred to inside `{}`. This is equivalent to the above, but it can be much nicer to work with.

```python
def greet(name):
    print(f"Hello, {name}")

greet('Amy')
```

### print vs. return

Functions often contain a `return` statement instead of (or in addition to) `print`. `print` just displays something — `return` gives the function's caller a value they can actually store and reuse.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
```

## Lists
A list is an ordered collection of values. We made one earlier (`fruits`). Each item has a position called an index, and indexing starts at 0, just like range().

```python
fruits = ['apple', 'banana', 'cherry']

print(fruits[0])    # first item
print(fruits[2])    # third item
print(fruits[-1])   # negative numbers count from the end
print(len(fruits))  # how many items
```

Lists can be changed after they're created, and you can check whether something is in one with `in`.

```python
fruits[1] = 'blueberry'   # replace the item at index 1
print(fruits)

print('apple' in fruits)  # True
print('grape' in fruits)  # False
```

## Dictionaries
A dictionary stores **key: value pairs**. Instead of looking things up by position (like a list), you look them up by name (the key). This shows up constantly in data work.

```python
student = {
    'name': 'Monda',
    'major': 'Computer Science',
    'year': 2
}

print(student['name'])   # look up a value by its key
```

You can add or update entries by assigning to a key, and loop over the keys with a `for` loop.

```python
student['year'] = 3          # update an existing key
student['gpa'] = 3.7         # add a new key

for key in student:
    print(key, '->', student[key])
```

We might want the key and value when we're looping through a dictionary. A neat way to do this is by using the `.items()` method.

```python
ages = {"Binta": 25, "Alexander": 30, "Isabella": 29}

for name, age in ages.items():
    print(f"{name} is {age} years old.")
```
