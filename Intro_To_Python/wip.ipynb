### Python Variable Naming Convetions

If you're familiar with other programming languages, you'll notice some naming habits differ here. Python doesn't actually enforce a style. You technically *can* name a variable almost anything and it'll run fine. But there are a few hard rules, and a lot of soft conventions that make life easier (for you and for whoever reads your code later).

**Hard rules (these will break your code):**
- Names can't start with a number (`2cool` is invalid, `cool2` is fine)
- No spaces or special characters like `-`, `!`, `@` (use `_` instead)
- Can't use reserved words like `if`, `for`, `return` as variable names

**Conventions:**
- Some languages use camelCase (`myVariableName`) — Python typically uses **snake_case** (`my_variable_name`)
- Constants (values that shouldn't change) are usually written in **ALL_CAPS**, like `MAX_SPEED = 120`

Try not to make variable names *too* long. You want something short enough to type comfortably, but descriptive enough that you can glance at it and immediately know what it holds.

You don't want to be writing out `number_of_students_in_my_school`, that's just a nightmare, but you also don't want to call it just `n`. Good middle spot is `num_students`.

### if, elif and else

`if` checks one condition. `elif` ("else if") checks another condition, but only if the first one was `False`. `else` runs if none of the conditions above it were `True`.

This also uses comparison operators: `>`, `<`, `==`, `!=`, `>=`, `<=`.

Notice how the assignment operator `=` is not the equality operator `==`.

```python
# assigning a variable
x = 5

# checking equality 
if x == 5:
    print('x is equal to 5')
```

```python
temperature = 75

if temperature > 90:
    print("It's hot")
elif temperature > 60:
    print("It's mild")
else:
    print("It's cold") # notice here that we use "" because we want to use ' inside the string
```

# Functions
A function is a reusable block of code that performs a specific task. Instead of rewriting the same lines every time you need to do something, you define it once and can call it as many times as you want, with different inputs.

```python
# functions

def greet(name):
    print("Hello,", name)

# enter your name
greet('Amy')
```

f-strings are formed like this: `f""`. Variables can be referred to inside `{}`. This is equivalent to the above, but it can be much nicer to work with.

```python
def greet(name):
    print(f"Hello, {name}")

greet('Amy')
```

### print vs. return

Functions often contain a `return` statement instead of (or in addition to) `print`. `print` just displays something — `return` gives the function's caller a value they can actually store and reuse.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
```
