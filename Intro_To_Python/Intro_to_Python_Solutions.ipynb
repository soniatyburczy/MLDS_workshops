## Mini-exercise: Predict the type

```python
print(type(7 + 2))   # int: adding two ints gives an int
print(type(7 / 2))   # float: division always returns a float
print(type('7'))     # str: it's in quotes, so it's text
```

## Mini-exercise: Letter grades

```python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("C")
```

Order matters here: if we checked `score >= 80` first, a 95 would print `"B"` because Python stops at the first condition that's `True`.

## Mini-exercise: Lists

```python
numbers = [4, 8, 15, 16, 23, 42]

print(numbers[0])     # first number
print(numbers[-1])    # last number
print(len(numbers))   # length

for number in numbers:
    print(number * 2)
```

## Final Exercise: Gradebook

```python
grades = {
    "Alice": 92,
    "Bob": 78,
    "Faith": 85,
    "Milo": 61
}

# Step 1: Write your function
def letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "C"

# Step 2: Loop through grades with .items()
for name, score in grades.items():
    print(f"{name} got a {score} ({letter_grade(score)})")
```

```python
# Step 3: Add a new student to grades, then run your loop again
grades["Sam"] = 88

for name, score in grades.items():
    print(f"{name} got a {score} ({letter_grade(score)})")
```

Notice that `letter_grade` **returns** the letter instead of printing it. That's what lets us drop its result right into the f-string.

```python
# Bonus: Write a class_average(grades) function
def class_average(grades):
    return sum(grades.values()) / len(grades)

print(class_average(grades))
```

`sum()` adds up all the scores and `len()` counts how many students there are. Dividing gives the average (and remember, division always returns a float).
